```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from crewai import LLM
llm = LLM(
    model="gemini/gemini-2.5-flash",
    temperature=0.1
)
llm.call("who invented the light bike?")
```

```python
original_email = "looping in vinay for TAS and PRX updates. Vinay, please share the latest updates on TAS and PRX with the team. and tell me if any blocker then ETA,let's sync up tommorow to discuss SYNCBOT"

from crewai.tools import BaseTool

class ReplaceJargonTools(BaseTool):
    name: str = "Replace Jargons"
    description: str = "This tool takes an email text and replaces technical jargons/abbreviations with their full forms. Returns the modified email with all jargons expanded."

    def _run(self, email: str) -> str:
        replacements = {
            "TAS": "Technical Assistance Service",
            "PRX": "Project Review Exchange",
            "ETA": "Estimated Time of Arrival",
            "SYNCBOT": "Synchronization Bot",
            "EOD": "End of Day",
        }
        result = email
        for jargon, replacement in replacements.items():
            result = result.replace(jargon, replacement)
        return result

jt = ReplaceJargonTools()
jt.run(original_email)
```

```python
from crewai import Agent, Crew, Task

email_assistant = Agent(
    role="Email Assistant",
    goal="Help users draft clear and concise professional emails based on their input.",
    backstory="You have been trained on a large dataset of emails and have a good understanding of email etiquette and structure.",
    tools=[jt],
    verbose=True,
    llm=llm
)

email_task = Task(
    description=f"""You are given an original email:
    '''{ original_email }'''
    
    Do the following:
    1. Use the Replace Jargons tool to expand all abbreviations in the email.
    2. Draft a professional, polished version of the email using the expanded text.""",
    agent=email_assistant,
    expected_output="A professional email with all abbreviations expanded and language polished."
)

crew = Crew(
    agents=[email_assistant],
    tasks=[email_task],
    verbose=True
)

results = await crew.kickoff_async()
print(results)
```
