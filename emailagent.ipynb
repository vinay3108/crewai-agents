```python
from dotenv import load_dotenv
import nest_asyncio
nest_asyncio.apply()
load_dotenv()
```

```python
from crewai import LLM
llm = LLM(
    model="gemini/gemini-2.5-flash",
    temperature=0.1
)
llm.call("who invented the light bulb?")
```

```python
from crewai import Agent, Crew,Task
email_assistant = Agent(
    role="You are an email assistant that helps users draft emails.",
    goal="Help users draft clear and concise emails based on their input.",
    backstory="You have been trained on a large dataset of emails and have a good understanding of email etiquette and structure.",
    verbose=True,
    llm=llm
)
original_email = "Hey, I need to talk to you about the project. Can we meet tomorrow before EOD?"

email_task = Task(
    description=f"""Draft a professional email based on the original email.",
    Expand Abbreviations: "Expand any abbreviations in the original email:
    '''{original_email}'''""",
    agent=email_assistant,
    expected_output="A professional email that conveys the same message as the original email, but in a more formal and polished manner."
)
crew = Crew(
    agents=[email_assistant],
    tasks=[email_task],
    verbose=True
)
results = await crew.kickoff_async()

print(results)
```
